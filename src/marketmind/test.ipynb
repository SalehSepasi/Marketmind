```python

import os
import requests
from dotenv import load_dotenv

load_dotenv()

BASE_URL = "https://api.marketaux.com/v1/news/all"


def fetch_news(
    topic: str,
    limit: int = 5,
    published_after: str | None = None,
) -> list[dict]:

    api_token = os.getenv("MARKETAUX_API_TOKEN")

    if not api_token:
        raise ValueError("MARKETAUX_API_TOKEN is not configured.")

    params = {
        "api_token": api_token,
        "search": topic,
        "language": "en",
        "limit": limit,
    }

    if published_after:
        params["published_after"] = published_after

    response = requests.get(
        BASE_URL,
        params=params,
        timeout=20,
    )
    response.raise_for_status()

    return response.json().get("data", [])


def main():
    # فقط حوزه موردنظر را تغییر بده
    topic = "cryptocurrency"

    articles = fetch_news(
        topic=topic,
        limit=5,
        published_after="2026-10-01",
    )

    print(f"\nTopic: {topic}")
    print(f"Number of articles: {len(articles)}")
    print("=" * 60)

    for i, article in enumerate(articles, start=1):
        print(f"\nArticle {i}")
        print("Title:", article.get("title"))
        print("Published:", article.get("published_at"))
        print("Source:", article.get("source"))
        print("Description:", article.get("description"))
        print("URL:", article.get("url"))
        print("-" * 60)


if __name__ == "__main__":
    main()
```
